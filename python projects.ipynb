```python
class QuizGame:
    def __init__(self):
        self.questions = [
            {
                "question": "What is the capital of India?",
                "options": ["A. Mumbai", "B. Delhi", "C. Chennai", "D. Kolkata"],
                "answer": "B"
            },
            {
                "question": "Which language is used in Python?",
                "options": ["A. English", "B. HTML", "C. Python", "D. Java"],
                "answer": "C"
            },
            {
                "question": "What is 5 + 3?",
                "options": ["A. 5", "B. 8", "C. 10", "D. 6"],
                "answer": "B"
            }
        ]
        self.score = 0

    def start(self):
        print("🎮 Welcome to Quiz Game!\n")

        for q in self.questions:
            print(q["question"])
            for opt in q["options"]:
                print(opt)

            user_ans = input("Enter your answer (A/B/C/D): ").upper()

            if user_ans == q["answer"]:
                print("✅ Correct!\n")
                self.score += 1
            else:
                print("❌ Wrong!\n")

        self.result()

    def result(self):
        print("🎯 Quiz Finished!")
        print("Your Score:", self.score, "/", len(self.questions))


# Run game
game = QuizGame()
game.start()
```

Availble books in library

```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
        self.avaliable = True

    def borrow_book(self):
        if self.avaliable:
            self.avaliable = False
            print(f"You have borrowed '{self.title}' by {self.author}.")
        else:
            print(f"Sorry, '{self.title}' is currently not available.")

    def return_book(self):
        if not self.avaliable:
            self.avaliable = True
            print(f"You have returned '{self.title}'. Thank you!")
        else:
            print(f"'{self.title}' was not borrowed.")

    def show_details(self):
        status = "Available" if self.avaliable else "Not Available"
        print(f"Title: {self.title}\nAuthor: {self.author}\nYear: {self.year}\nStatus: {status}")

book1 = Book("life of me", "Harper Lee", 1960)
book1.show_details()   
book1.borrow_book() 
book2= Book("1984", "George Orwell", 1949)
book2.show_details()
book2.borrow_book()
```

Library code

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.available = True

    def borrow_book(self):
        if self.available:
            self.available = False
            print("Book borrowed successfully")
        else:
            print("Book is already borrowed")

    def return_book(self):
        self.available = True
        print("Book returned successfully")

    def display(self):
        print("Title:", self.title)
        print("Author:", self.author)
        print("Available:", self.available)
        print("-------------------")


class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)
        print(book.title, "added to library")

    def show_books(self):
        print("\nLibrary Books:")
        for book in self.books:
            book.display()


# Creating books
book1 = Book("Python Basics", "Manu")
book2 = Book("Java Basics", "Ravi")

# Creating library
library = Library()

# Adding books
library.add_book(book1)
library.add_book(book2)

# Show books
library.show_books()

# Borrow a book
book1.borrow_book()

# Show books after borrowing
library.show_books()

# Return book
book1.return_book()

# Show books after returning
library.show_books()
```
